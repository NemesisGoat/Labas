```python
import numpy as np
import sympy as sp

fio = "Клочко Роман Андреевич"
N = sum(1 for c in fio if c.isalpha())
variant = N % 15
print("ФИО:", fio)
print("Количество букв N =", N)
print("Вариант =", variant)

a, b = -0.2, 0.4

def f(x):
    return (0.1 + x**2) * np.sin(x) / (x**2 + 0.5)

def f_sym(x):
    return (sp.Rational(1,10) + x**2) * sp.sin(x) / (x**2 + sp.Rational(1,2))

print("Задача 1. Вычисление интеграла с точностью eps = 10^-4 методом составных трапеций")
print("Интервал [a, b] = [", a, ",", b, "]")
print("f(x) = (0.1 + x^2) * sin(x) / (x^2 + 0.5)")

x_sym = sp.Symbol('x')
exact = sp.integrate(f_sym(x_sym), (x_sym, a, b))
exact_val = float(exact)
print("Точное значение интеграла (символьное):", exact)
print("Точное значение интеграла (численное):", exact_val)

eps = 1e-4

def trapezoid(f, a, b, n):
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    y = f(x)
    result = h * (0.5 * y[0] + np.sum(y[1:-1]) + 0.5 * y[-1])
    return result, h, x, y

n = 1
trap_val, h, x, y = trapezoid(f, a, b, n)
print("Шаг 1: n =", n, ", h =", h)
print("Узлы x:", x)
print("Значения f(x):", y)
print("Формула: h * (0.5*f(x0) + f(x1) + ... + 0.5*f(xn))")
print("Результат:", trap_val)
print("Погрешность:", abs(trap_val - exact_val))
print()

while abs(trap_val - exact_val) > eps:
    n *= 2
    trap_val, h, x, y = trapezoid(f, a, b, n)
    print("Шаг: n =", n, ", h =", h)
    print("Узлы x:", x)
    print("Значения f(x):", y)
    print("Результат:", trap_val)
    print("Погрешность:", abs(trap_val - exact_val))
    print()

print("Ответ задачи 1:")
print("n =", n, ", h =", h)
print("Значение интеграла:", trap_val)
print("Погрешность:", abs(trap_val - exact_val))
print()

print("Задача 2. Вычисление интеграла по квадратурной формуле Гаусса с тремя узлами")
print("Формула Гаусса: интеграл от -1 до 1 g(t)dt приближается суммой w_i * g(t_i)")
print("Узлы t_i и веса w_i для 3 узлов:")
print("t1 = -sqrt(3/5), w1 = 5/9")
print("t2 = 0, w2 = 8/9")
print("t3 = sqrt(3/5), w3 = 5/9")

t_nodes = [-np.sqrt(3/5), 0, np.sqrt(3/5)]
weights = [5/9, 8/9, 5/9]

print("Узлы на [-1, 1]:", t_nodes)
print("Веса:", weights)

print("Замена переменной: x = (b+a)/2 + (b-a)/2 * t")
print("a =", a, ", b =", b)
print("(b+a)/2 =", (b+a)/2)
print("(b-a)/2 =", (b-a)/2)

gauss_val = 0
print("Вычисление:")
for i in range(3):
    t = t_nodes[i]
    w = weights[i]
    x_transformed = (b + a) / 2 + (b - a) / 2 * t
    fx = f(x_transformed)
    term = w * fx
    gauss_val += term
    print("i =", i+1)
    print("  t =", t)
    print("  x = (b+a)/2 + (b-a)/2 * t =", x_transformed)
    print("  f(x) =", fx)
    print("  w =", w)
    print("  w * f(x) =", term)

gauss_val = (b - a) / 2 * gauss_val
print("Сумма w_i * f(x_i) =", gauss_val / ((b - a) / 2))
print("Умножаем на (b-a)/2 =", (b - a) / 2)
print("Результат Гаусса:", gauss_val)
print("Погрешность Гаусса:", abs(gauss_val - exact_val))
print()

print("Сравнение:")
print("Точное значение:", exact_val)
print("Трапеции n=%d:" % n, trap_val, "погрешность", abs(trap_val - exact_val))
print("Гаусс 3 узла:", gauss_val, "погрешность", abs(gauss_val - exact_val))
```

```python
import numpy as np
import sympy as sp

print("=" * 60)
print("ЗАДАЧА 1: Формула Симпсона")
print("=" * 60)

f_values = {0: -1, 0.5: 1, 1: 2.5}
a, b = 0, 1

print("\nДано:")
print(f"  f(0) = {f_values[0]}")
print(f"  f(1/2) = {f_values[0.5]}")
print(f"  f(1) = {f_values[1]}")

print("\nШаг 1: Определяем g(x) = x^2 * f(x)")
g0 = 0**2 * f_values[0]
g_half = (0.5)**2 * f_values[0.5]
g1 = 1**2 * f_values[1]

print(f"  g(0) = 0^2 * f(0) = 0 * ({f_values[0]}) = {g0}")
print(f"  g(1/2) = (1/2)^2 * f(1/2) = 0.25 * {f_values[0.5]} = {g_half}")
print(f"  g(1) = 1^2 * f(1) = 1 * {f_values[1]} = {g1}")

print("\nШаг 2: Применяем формулу Симпсона")
print(f"  интеграл = (b-a)/6 * [g(a) + 4*g((a+b)/2) + g(b)]")
print(f"  интеграл = 1/6 * [{g0} + 4*{g_half} + {g1}]")

result1 = (b - a) / 6 * (g0 + 4*g_half + g1)

print(f"  интеграл = 1/6 * {g0 + 4*g_half + g1}")
print(f"  интеграл = {result1}")

x = sp.Symbol('x')
x_vals = [0, sp.Rational(1,2), 1]
y_vals = [-1, 1, sp.Rational(5,2)]

L = 0
for i in range(3):
    term = y_vals[i]
    for j in range(3):
        if i != j:
            term *= (x - x_vals[j]) / (x_vals[i] - x_vals[j])
    L += term

L_simplified = sp.simplify(L)
exact1 = sp.integrate(x**2 * L_simplified, (x, 0, 1))

print(f"\nШаг 3: Проверка (интерполяционный многочлен)")
print(f"  L(x) = {L_simplified}")
print(f"  Точное значение = {exact1} = {float(exact1)}")
print(f"  Погрешность Симпсона = {abs(result1 - float(exact1))}")

print("\n" + "=" * 60)
print("ЗАДАЧА 2: Погрешность средних прямоугольников")
print("=" * 60)

a2, b2 = -1, 0

print(f"\nДано:")
print(f"  f(x) = sin(x)")
print(f"  a = {a2}, b = {b2}")

x2 = sp.Symbol('x')
exact2 = sp.integrate(sp.sin(x2), (x2, a2, b2))
exact2_val = float(exact2)

print(f"\nШаг 1: Точное значение интеграла")
print(f"  = [-cos(x)] от {a2} до {b2}")
print(f"  = -1 + cos(1) = {exact2_val}")

print(f"\nШаг 2: Формула средних прямоугольников")
h2 = b2 - a2
x_mid2 = (a2 + b2) / 2
approx2 = h2 * np.sin(x_mid2)

print(f"  h = {h2}")
print(f"  x_mid = {x_mid2}")
print(f"  f(x_mid) = sin({x_mid2}) = {np.sin(x_mid2)}")
print(f"  Приближенное = h * f(x_mid) = {approx2}")

print(f"\nШаг 3: Теоретическая оценка погрешности")
print(f"  |R| <= (b-a)^3 / 24 * max|f''(x)|")

f2_prime = sp.diff(sp.sin(x2), x2, 2)
print(f"  f''(x) = {f2_prime}")

max_f2 = np.sin(1)
error_bound2 = (b2 - a2)**3 / 24 * max_f2

print(f"  max|f''(x)| = sin(1) = {max_f2}")
print(f"  |R| <= 1/24 * {max_f2} = {error_bound2}")

actual_error2 = abs(approx2 - exact2_val)
print(f"\nШаг 4: Фактическая погрешность")
print(f"  |{exact2_val} - {approx2}| = {actual_error2}")
print(f"  Оценка верна: {actual_error2 <= error_bound2}")

print("\n" + "=" * 60)
print("ЗАДАЧА 3: Узлы и коэффициенты Гауссовой квадратуры")
print("=" * 60)

a3, b3 = 0, 1
n3 = 2

print(f"\nДано:")
print(f"  интеграл от {a3} до {b3} f(x)dx = A0*f(x0) + A1*f(x1)")
print(f"  n = {n3}")

print(f"\nШаг 1: Метод моментов")
x3 = sp.Symbol('x')
moments = []
for k in range(4):
    moment = sp.integrate(x3**k, (x3, a3, b3))
    moments.append(moment)
    print(f"  интеграл x^{k} dx = {moment}")

print(f"\nШаг 2: Нахождение узлов")
print(f"  Узлы Гаусса - корни многочлена Лежандра степени n=2")

t3 = sp.Symbol('t')
P2 = (3*t3**2 - 1)/2
roots_t = sp.solve(P2, t3)

print(f"  P2(t) = {P2}")
print(f"  Корни:")
for i, root in enumerate(roots_t):
    print(f"  t{i} = {root} = {float(root)}")

print(f"\n  Замена: x = (t+1)/2")
nodes3 = []
for i, root in enumerate(roots_t):
    x_node = (root + 1)/2
    nodes3.append(x_node)
    print(f"  x{i} = ({root}+1)/2 = {x_node} = {float(x_node)}")

print(f"\nШаг 3: Нахождение коэффициентов")
A0, A1 = sp.symbols('A0 A1')
eq1 = sp.Eq(A0 + A1, 1)
eq2 = sp.Eq(A0*nodes3[0] + A1*nodes3[1], sp.Rational(1,2))

print(f"  Система:")
print(f"  {eq1}")
print(f"  {eq2}")

solution3 = sp.solve([eq1, eq2], [A0, A1])
A0_val = solution3[A0]
A1_val = solution3[A1]

print(f"\n  Решение:")
print(f"  A0 = {A0_val} = {float(A0_val)}")
print(f"  A1 = {A1_val} = {float(A1_val)}")

print(f"\nШаг 4: Проверка на всех моментах")
for k in range(4):
    lhs = moments[k]
    rhs = A0_val * nodes3[0]**k + A1_val * nodes3[1]**k
    rhs_simplified = sp.simplify(rhs)
    check = sp.simplify(lhs - rhs_simplified) == 0
    print(f"  x^{k}: {lhs} = {rhs_simplified} ? {check}")

print(f"\nШаг 5: Итоговая формула")
print(f"  интеграл от 0 до 1 f(x)dx = {A0_val}*f({nodes3[0]}) + {A1_val}*f({nodes3[1]})")
print(f"\n  Численно:")
print(f"  = {float(A0_val)}*f({float(nodes3[0])}) + {float(A1_val)}*f({float(nodes3[1])})")
```
